6. Explain the concept of generators in Python and how they are defined?
a Generator Function Is a User-Defined Function That Contains Yield Instead Of return Statemetn In Same User Defined Function.  
def Gen_Func(Variable_Name):
	for i in Variable_Name:
		yield i

a = Gen_Func([1,2,3,4,5,6])
print(next(a)) -> 1
print(next(a)) -> 2 ….And So On
Generator Object Gives One Element, Exits The Iteration And Keeps Its State In Mind Where It Has Stopped. When We Use next() Again, It Resumes Execution, Gives The Next Value, And Pauses Again Until The Next next() Call.
Generators are very useful with large datasets, as with a generator function, instead of loading the whole dataset into memory, we can generate and process one value at a time

```python
"""
Practical Questions
1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.
"""
def sum_num(number):
  total = 0
  for num in number:
    if num % 2  == 0:
      total += num
  return total

numbers = [12, 7, 24, 15, 8, 31, 42, 19, 6, 25, 10, 33, 18, 5, 20]
print(sum_num(numbers))
```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.
Prime_Number = []

for i in range(2,201):
  Dividing_Number = []
  for j in range(2, i+1):
    if i % j == 0:
      Dividing_Number.append(j)
  if len(Dividing_Number) <= 1:
    Prime_Number.append(i)

print(f'Prime Numbers: {Prime_Number}')
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms
def Fic_Func(N):
  a, b = 0, 1
  for i in range(N):
    yield a
    a, b = b, a + b

for num in Fic_Func(10):
  print(num)
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
def power_func():
  n = int(input('Enter Your Number Here: '))
  for i in range(1,n+1):
    yield 2 ** i

Ans = power_func()
for value in Ans:
  print(value)
```

```python
# 9  Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit
import math as m

celsius = [0, 10, 20, 25, 30, 37, 40]

func = lambda x: (x*1.8) + 32

answer = (list(map(func,celsius)))
for i in answer:
  print(m.ceil(i))
```
